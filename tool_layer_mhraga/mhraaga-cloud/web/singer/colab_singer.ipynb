# mhRaaga singer: make the raga song SING with YuE2 (free Colab GPU)

Runs the open **YuE2** model on Colab's free GPU. Set the runtime to **GPU** first (Runtime > Change runtime type).

- It sings **our raga melody** (the score came from mhRaaga), with your lyrics, in the style written in the request.
- Two versions are made so you can hear which spelling the model sings better: Devanagari and romanised Hindi.
- **Honest limits:** the free T4 has 16 GB, the model prefers 24 GB, so this uses its low-memory mode and may run out of memory; if so, use a bigger runtime or fewer lyric lines. YuE2 does not list Hindi among its trained languages, so pronunciation may be poor.
- **Licence:** YuE2 weights are free for personal and creator use and for non-commercial work; a company needs a licence (see third_party/YuE/README.md).

```
import time, json, pathlib
from yue2 import YuE2Pipeline
pathlib.Path('out').mkdir(exist_ok=True)
start = time.time()
with YuE2Pipeline.from_pretrained('m-a-p/YuE2-3B', device='cuda', low_vram=True) as pipe:
    for r in REQUESTS:
        req = {k: r[k] for k in ('id', 'style', 'lyrics', 'abc', 'cot', 'seed')}
        try:
            song = pipe(**req)
            song.save_artifacts('out/' + r['id'])
            print(r['id'], 'done, truncated =', song.truncated, round(time.time() - start), 's')
        except Exception as e:
            print(r['id'], 'FAILED:', type(e).__name__, str(e)[:300])
```

```
import glob, subprocess
from google.colab import files
for f in sorted(glob.glob('out/*/audio.*')):
    mp3 = f.rsplit('/', 1)[0] + '.mp3'
    subprocess.run(['ffmpeg', '-y', '-v', 'error', '-i', f, '-b:a', '192k', mp3], check=True)
    files.download(mp3)
```
